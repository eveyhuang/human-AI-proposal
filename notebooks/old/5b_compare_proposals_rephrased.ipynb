```python

from itertools import combinations
from collections import Counter
import ast
import pickle

from IPython.display import display
from scipy.stats import mannwhitneyu, kruskal, fisher_exact, chi2_contingency
from sklearn.decomposition import LatentDirichletAllocation
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_distances
from statsmodels.stats.multitest import multipletests

GROUP_ORDER = ['Human', 'Claude', 'Gemini', 'GPT']
MODEL_ORDER = ['Claude', 'Gemini', 'GPT']
GROUP_PALETTE = {
    'Human': '#DC143C',
    'Claude': '#4A90E2',
    'Gemini': '#7B68EE',
    'GPT': '#3CB371',
    'All AI': '#4A90E2',
    'All-AI-baseline': '#4A90E2',
    'All-AI-one_at_a_time': '#7B68EE',
    'All-AI-persona': '#3CB371',
    'all-ai-baseline': '#4A90E2',
    'all-ai-one-at-a-time': '#7B68EE',
    'all-ai-persona': '#3CB371',
}
CONDITION_LABELS = {
    'baseline': 'all-ai-baseline',
    'one_at_a_time': 'all-ai-one-at-a-time',
    'persona': 'all-ai-persona',
}
NOVELTY_METRICS = [
    'literature_element_novelty_k1',
    'literature_mean_knn_novelty_k10',
    'literature_local_density_normalized_novelty',
]
AUTHOR_LOCAL_METRICS = ['author_group_mean_pairwise_distance', 'author_group_nearest_neighbor_distance']
NN_METRICS = ['global_nn_distance', 'global_mean_5nn_distance']
BERTOPIC_METRICS = ['bertopic_region_entropy', 'bertopic_max_region_weight', 'bertopic_unique_regions']
MESH_METRICS = ['mesh_unique_terms_k20']
MATRIX_DERIVED_METRICS = [
    'mean_pairwise_distance',
    'nearest_neighbor_distance',
    *AUTHOR_LOCAL_METRICS,
    *NN_METRICS,
]
INDEPENDENT_PROPOSAL_METRICS = [
    *NOVELTY_METRICS,
    *BERTOPIC_METRICS,
    *MESH_METRICS,
]
PROPOSAL_LEVEL_METRICS = [
    *MATRIX_DERIVED_METRICS,
    *INDEPENDENT_PROPOSAL_METRICS,
]

LIT_ARTICLE_INDEX_PATH = PROJECT_ROOT / 'data' / 'prepared' / 'literature' / 'literature_article_index.csv'
LIT_BERTOPIC_ASSIGNMENTS_PATH = PROJECT_ROOT / 'data' / 'prepared' / 'rephrased' / 'minimal' / 'lit_bertopic_assignments.csv'
LIT_BERTOPIC_TOPIC_INFO_PATH = PROJECT_ROOT / 'data' / 'prepared' / 'rephrased' / 'minimal' / 'lit_bertopic_topic_info.csv'
LIT_UMAP2D_PATH = PROJECT_ROOT / 'data' / 'embeddings' / 'literature' / 'lit_umap2d.npy'
LIT_UMAP_REDUCER_PATH = PROJECT_ROOT / 'data' / 'embeddings' / 'literature' / 'lit_umap_reducer.pkl'


def p_stars(p):
    if pd.isna(p):
        return ''
    if p < 0.001:
        return '***'
    if p < 0.01:
        return '**'
    if p < 0.05:
        return '*'
    return ''


def cliffs_delta(x, y):
    x = np.asarray(pd.Series(x).dropna(), dtype=float)
    y = np.asarray(pd.Series(y).dropna(), dtype=float)
    if len(x) == 0 or len(y) == 0:
        return np.nan
    comp = x[:, None] - y[None, :]
    return float((np.sum(comp > 0) - np.sum(comp < 0)) / comp.size)


def bootstrap_mean_ci(values, n_boot=1000, seed=42):
    values = np.asarray(pd.Series(values).dropna(), dtype=float)
    if len(values) == 0:
        return np.nan, np.nan, np.nan
    rng = np.random.default_rng(seed)
    boot = np.array([rng.choice(values, size=len(values), replace=True).mean() for _ in range(n_boot)])
    return float(values.mean()), float(np.percentile(boot, 2.5)), float(np.percentile(boot, 97.5))


def holm_q(pvals):
    pvals = pd.Series(pvals, dtype=float)
    out = pd.Series(np.nan, index=pvals.index, dtype=float)
    mask = pvals.notna()
    if mask.any():
        out.loc[mask] = multipletests(pvals.loc[mask].to_numpy(), method='holm')[1]
    return out


def truthy_flag(value):
    if pd.isna(value):
        return False
    if isinstance(value, str):
        return value.strip().lower() in {'true', '1', 'yes', 'accepted', 'funded'}
    return bool(value)


def group_n_map(df, group_col, groups=None, *, count_col=None):
    if groups is None:
        groups = [g for g in pd.Series(df[group_col]).dropna().unique()]
    if count_col and count_col in df.columns:
        counts = df.groupby(group_col)[count_col].first().to_dict()
    else:
        counts = df.groupby(group_col).size().to_dict()
    return {g: int(counts.get(g, 0)) for g in groups}


def label_with_n(group, n_map=None):
    if n_map is None or group not in n_map or pd.isna(n_map.get(group)):
        return str(group)
    return f'{group}\n(N={int(n_map[group])})'


def apply_group_n_xticklabels(ax, groups, n_map=None, rotation=20):
    ax.set_xticks(np.arange(len(groups)))
    ax.set_xticklabels([label_with_n(g, n_map) for g in groups], rotation=rotation)


def relabel_index_with_n(table_df, n_map=None):
    if n_map is None:
        return table_df
    out = table_df.copy()
    out.index = [label_with_n(idx, n_map) for idx in out.index]
    return out


def relabel_legend_with_n(ax, n_map=None, *, title=None):
    handles, labels = ax.get_legend_handles_labels()
    if not handles:
        return
    new_labels = []
    for label in labels:
        clean = str(label)
        new_labels.append(label_with_n(clean, n_map) if n_map and clean in n_map else clean)
    ax.legend(handles, new_labels, title=title, loc='upper left', bbox_to_anchor=(1.02, 1.0), frameon=False, fontsize=9)


def pairwise_mw_tests(df, value_col, group_col='source_group', groups=None, condition=None, family=None):
    groups = groups or [g for g in GROUP_ORDER if g in set(df[group_col].dropna())]
    rows = []
    for g1, g2 in combinations(groups, 2):
        x = pd.to_numeric(df.loc[df[group_col] == g1, value_col], errors='coerce').dropna()
        y = pd.to_numeric(df.loc[df[group_col] == g2, value_col], errors='coerce').dropna()
        if len(x) < 2 or len(y) < 2:
            continue
        stat, p = mannwhitneyu(x, y, alternative='two-sided')
        rows.append({
            'condition': condition,
            'analysis_family': family,
            'metric': value_col,
            'group_1': g1,
            'group_2': g2,
            'n_1': len(x),
            'n_2': len(y),
            'mean_1': float(x.mean()),
            'mean_2': float(y.mean()),
            'difference_1_minus_2': float(x.mean() - y.mean()),
            'cliffs_delta_1_minus_2': cliffs_delta(x, y),
            'mannwhitney_u': float(stat),
            'p_value': float(p),
            'test': 'Mann-Whitney U',
            'p_adjustment': 'Holm',
        })
    out = pd.DataFrame(rows)
    if len(out):
        out['q_value_holm'] = holm_q(out['p_value']).to_numpy()
        out['significance'] = out['q_value_holm'].map(p_stars)
    return out


def kruskal_group_test(df, value_col, group_col='source_group', groups=None, condition=None, family=None):
    groups = groups or [g for g in GROUP_ORDER if g in set(df[group_col].dropna())]
    samples = [pd.to_numeric(df.loc[df[group_col] == g, value_col], errors='coerce').dropna().to_numpy() for g in groups]
    samples = [s for s in samples if len(s) >= 2]
    if len(samples) < 2:
        return {'condition': condition, 'analysis_family': family, 'metric': value_col, 'test': 'Kruskal-Wallis', 'p_value': np.nan}
    stat, p = kruskal(*samples)
    return {'condition': condition, 'analysis_family': family, 'metric': value_col, 'test': 'Kruskal-Wallis', 'statistic': float(stat), 'p_value': float(p), 'n_groups': len(samples), 'p_adjustment': 'Holm'}


def save_show(fig, path):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    fig.tight_layout()
    fig.savefig(path, dpi=220, bbox_inches='tight')
    display(fig)
    plt.close(fig)


def stable_point_jitter(row, *, width=0.16):
    key = str(row.get('proposal_uid', row.name))
    seed = sum((i + 1) * ord(ch) for i, ch in enumerate(key)) % (2**32 - 1)
    return np.random.default_rng(seed + BOOT_SEED).uniform(-width, width)


def add_metadata_rings(ax, plot_df, value_col, x_col='_plot_x'):
    # Rings reuse the exact x coordinate used for the visible point layer.
    if x_col not in plot_df.columns:
        return
    for _, row in plot_df.iterrows():
        if pd.isna(row.get(value_col)) or pd.isna(row.get(x_col)):
            continue
        x = row[x_col]
        y = row[value_col]
        if truthy_flag(row.get('is_funded_human', False)):
            ax.scatter([x], [y], s=72, facecolors='none', edgecolors='#FF00FF', linewidths=1.5, zorder=8)
        if truthy_flag(row.get('is_top5_AI-ranked', False)) or truthy_flag(row.get('is_top5_AI_ranked', False)):
            ax.scatter([x], [y], s=104, facecolors='none', edgecolors='#FF8C00', linewidths=1.9, zorder=9)


def _test_label_from_frame(frame):
    if frame is None or frame.empty:
        return 'descriptive only'
    tests = [str(x) for x in frame.get('test', pd.Series(dtype=object)).dropna().unique() if str(x).strip()]
    if not tests:
        if 'mannwhitney_u' in frame.columns:
            tests = ['Mann-Whitney U']
        elif 'permutation_p_value' in frame.columns:
            tests = ['label permutation']
    adjustment = [str(x) for x in frame.get('p_adjustment', pd.Series(dtype=object)).dropna().unique() if str(x).strip()]
    label = ' / '.join(tests) if tests else 'test not named'
    if adjustment:
        label = f'{label}; {" / ".join(adjustment)}'
    return label


def _forest_x_column(frame):
    for col in ['cliffs_delta_1_minus_2', 'effect_1_minus_2', 'difference_1_minus_2']:
        if col in frame.columns:
            return col
    return None


def _forest_x_label(x_col):
    if x_col == 'cliffs_delta_1_minus_2':
        return "Cliff's delta (group 1 - group 2)"
    if x_col == 'effect_1_minus_2':
        return 'Group statistic difference (group 1 - group 2)'
    if x_col == 'difference_1_minus_2':
        return 'Mean difference (group 1 - group 2)'
    return 'Effect size'


def plot_box_ci_forest(df, value_col, *, group_col='source_group', groups=None, title='', out_path=None, tests_df=None, n_map=None):
    groups = groups or [g for g in GROUP_ORDER if g in set(df[group_col].dropna())]
    plot_df = df[df[group_col].isin(groups)].copy()
    plot_df[value_col] = pd.to_numeric(plot_df[value_col], errors='coerce')
    plot_df = plot_df.dropna(subset=[value_col])
    if plot_df.empty:
        print(f'No data to plot for {value_col}')
        return
    n_map = n_map or group_n_map(plot_df, group_col, groups)
    palette = {g: GROUP_PALETTE.get(g, '#808080') for g in groups}
    fig, axes = plt.subplots(1, 2, figsize=(15, 5.2), gridspec_kw={'width_ratios': [2.2, 1.55]})
    ax = axes[0]
    sns.boxplot(data=plot_df, x=group_col, y=value_col, order=groups, hue=group_col, palette=palette, ax=ax, width=0.55, fliersize=0, linewidth=1.1, legend=False)
    x_positions = {g: i for i, g in enumerate(groups)}
    plot_df['_plot_x'] = plot_df.apply(lambda row: x_positions[row[group_col]] + stable_point_jitter(row), axis=1)
    point_colors = plot_df[group_col].map(lambda g: palette.get(g, '#808080'))
    ax.scatter(plot_df['_plot_x'], plot_df[value_col], s=18, c=point_colors, alpha=0.55, edgecolors='white', linewidths=0.3, zorder=4)
    for g in groups:
        vals = plot_df.loc[plot_df[group_col] == g, value_col]
        mean, lo, hi = bootstrap_mean_ci(vals, seed=BOOT_SEED)
        if pd.notna(mean):
            ax.errorbar(x_positions[g], mean, yerr=[[mean - lo], [hi - mean]], fmt='D', color='black', mfc=palette[g], mec='black', ms=7, capsize=4, zorder=7)
    add_metadata_rings(ax, plot_df, value_col)
    ax.set_title(title or value_col)
    ax.set_xlabel('Author group' if group_col == 'source_group' else group_col)
    ax.set_ylabel(value_col)
    apply_group_n_xticklabels(ax, groups, n_map, rotation=22)

    axf = axes[1]
    if tests_df is None:
        tests_df = pairwise_mw_tests(plot_df, value_col, group_col=group_col, groups=groups)
    forest = tests_df[(tests_df['metric'] == value_col) & (tests_df['group_1'] == groups[0])].copy() if len(tests_df) else pd.DataFrame()
    if forest.empty and len(tests_df):
        forest = tests_df[tests_df['metric'] == value_col].copy().head(12)
        forest['label'] = forest['group_1'].astype(str) + ' vs ' + forest['group_2'].astype(str)
    elif not forest.empty:
        forest['label'] = forest['group_2'].astype(str)
    x_col = _forest_x_column(forest) if not forest.empty else None
    if not forest.empty and x_col is not None:
        forest[x_col] = pd.to_numeric(forest[x_col], errors='coerce')
        forest = forest.dropna(subset=[x_col]).sort_values(x_col)
    if not forest.empty and x_col is not None:
        p_col = 'q_value_holm' if 'q_value_holm' in forest.columns else ('p_value' if 'p_value' in forest.columns else None)
        colors = []
        for _, r in forest.iterrows():
            sig_p = r.get(p_col, np.nan) if p_col else np.nan
            group_for_color = r.get('group_2', r.get('label', ''))
            colors.append('#666666' if pd.isna(sig_p) or sig_p >= 0.05 else GROUP_PALETTE.get(group_for_color, '#333333'))
        axf.axvline(0, color='black', linewidth=1)
        axf.scatter(forest[x_col], np.arange(len(forest)), c=colors, s=70)
        for yi, (_, r) in enumerate(forest.iterrows()):
            star = p_stars(r.get(p_col, np.nan)) if p_col else ''
            if star:
                axf.text(r[x_col], yi + 0.14, star, ha='center', va='bottom', fontsize=10)
        axf.set_yticks(np.arange(len(forest)))
        axf.set_yticklabels(forest['label'])
        xmax = float(np.nanmax(np.abs(forest[x_col]))) if len(forest) else 1.0
        xmax = max(0.05, xmax * 1.25)
        if x_col == 'cliffs_delta_1_minus_2':
            axf.set_xlim(-1.05, 1.05)
        else:
            axf.set_xlim(-xmax, xmax)
        axf.set_xlabel(_forest_x_label(x_col))
        axf.set_title(f'Effect panel: {_test_label_from_frame(forest)}')
    else:
        axf.axis('off')
        axf.text(0.5, 0.5, 'No valid pairwise test for this metric', ha='center', va='center', wrap=True)
    handles = [plt.Line2D([0], [0], marker='D', color='black', markerfacecolor='white', linestyle='', label='Mean with bootstrap 95% CI'),
               plt.Line2D([0], [0], marker='o', color='#FF00FF', markerfacecolor='none', linestyle='', label='Accepted/Funded Human ring'),
               plt.Line2D([0], [0], marker='o', color='#FF8C00', markerfacecolor='none', linestyle='', label='Top AI-ranked Human ring')]
    ax.legend(handles=handles, loc='upper left', bbox_to_anchor=(1.02, 1.0), frameon=False, fontsize=9)
    if out_path:
        save_show(fig, out_path)


def plot_stacked_counts(df, row_col, col_col, *, title, out_path, normalize=True, palette=None, n_map=None):
    tab = pd.crosstab(df[row_col], df[col_col], normalize='index' if normalize else False)
    n_map = n_map or group_n_map(df, row_col, list(tab.index))
    ordered_rows = [g for g in n_map.keys() if g in tab.index] + [g for g in tab.index if g not in n_map]
    tab = tab.reindex(ordered_rows)
    fig, ax = plt.subplots(figsize=(10, 5))
    colors = palette or sns.color_palette('tab20', n_colors=tab.shape[1])
    tab.plot(kind='bar', stacked=True, ax=ax, color=colors, width=0.75)
    ax.set_title(title)
    ax.set_ylabel('Proportion' if normalize else 'Count')
    ax.set_xlabel(row_col)
    ax.set_xticklabels([label_with_n(g, n_map) for g in tab.index], rotation=20)
    ax.legend(title=col_col, loc='upper left', bbox_to_anchor=(1.02, 1.0), frameon=False, fontsize=9)
    save_show(fig, out_path)


def plot_weighted_stacked(df, row_col, col_col, weight_col, *, title, out_path, top_n=10, n_map=None):
    if df.empty:
        print(f'No weighted data for {title}')
        return
    work = df.copy()
    top_cols = work.groupby(col_col)[weight_col].sum().sort_values(ascending=False).head(top_n).index
    work[col_col] = np.where(work[col_col].isin(top_cols), work[col_col], 'Other')
    tab = work.pivot_table(index=row_col, columns=col_col, values=weight_col, aggfunc='sum', fill_value=0)
    tab = tab.div(tab.sum(axis=1).replace(0, np.nan), axis=0).fillna(0)
    n_map = n_map or group_n_map(df, row_col, list(tab.index))
    ordered_rows = [g for g in n_map.keys() if g in tab.index] + [g for g in tab.index if g not in n_map]
    tab = tab.reindex(ordered_rows)
    fig, ax = plt.subplots(figsize=(11, 5))
    tab.plot(kind='bar', stacked=True, ax=ax, colormap='tab20', width=0.75)
    ax.set_title(title)
    ax.set_ylabel('Mean neighbor weight')
    ax.set_xlabel(row_col)
    ax.set_xticklabels([label_with_n(g, n_map) for g in tab.index], rotation=20)
    ax.legend(title=col_col, loc='upper left', bbox_to_anchor=(1.02, 1.0), frameon=False, fontsize=8)
    save_show(fig, out_path)


def plot_heatmap(table_df, *, title, out_path, fmt='.2f', cmap='viridis'):
    fig, ax = plt.subplots(figsize=(max(7, table_df.shape[1] * 0.9), max(4, table_df.shape[0] * 0.45)))
    sns.heatmap(table_df, annot=True, fmt=fmt, cmap=cmap, ax=ax, cbar_kws={'shrink': 0.75})
    ax.set_title(title)
    save_show(fig, out_path)


def run_fixed_lda(master_df, *, n_topics=5, seed=42):
    texts = master_df['abstract_text'].fillna('').astype(str)
    vectorizer = CountVectorizer(stop_words='english', min_df=2)
    X = vectorizer.fit_transform(texts)
    lda = LatentDirichletAllocation(n_components=n_topics, random_state=seed)
    probs = lda.fit_transform(X)
    topics = probs.argmax(axis=1)
    out = master_df[['proposal_uid', 'source_type', 'source_group']].copy()
    out['lda_topic'] = topics
    for k in range(n_topics):
        out[f'lda_topic_prob_{k}'] = probs[:, k]
    words = np.asarray(vectorizer.get_feature_names_out())
    labels = []
    for topic_idx, comp in enumerate(lda.components_):
        top_words = words[np.argsort(comp)[::-1][:8]].tolist()
        labels.append({'lda_topic': topic_idx, 'top_terms': '; '.join(top_words)})
    test_rows = []
    for grouping in ['source_type', 'source_group']:
        tab = pd.crosstab(out[grouping], out['lda_topic'])
        chi2, p, dof, expected = chi2_contingency(tab)
        n = tab.to_numpy().sum()
        cramers_v = np.sqrt(chi2 / (n * max(1, min(tab.shape) - 1)))
        test_rows.append({'grouping': grouping, 'chi2': float(chi2), 'p_value': float(p), 'dof': int(dof), 'cramers_v': float(cramers_v), 'n_topics': n_topics})
    return out, pd.DataFrame(test_rows), pd.DataFrame(labels)


def _global_knn_group_mean(D, idx, *, k=1):
    if len(idx) == 0:
        return np.nan
    arr = np.asarray(D[np.ix_(idx, np.arange(D.shape[0]))], dtype=float).copy()
    for row_pos, global_idx in enumerate(idx):
        arr[row_pos, global_idx] = np.inf
    k_eff = min(k, arr.shape[1] - 1)
    if k_eff <= 0:
        return np.nan
    nearest = np.partition(arr, kth=k_eff - 1, axis=1)[:, :k_eff]
    return float(np.mean(nearest))


def _matrix_group_stat(D, idx, metric_name, *, coords=None, coord_range=None):
    if metric_name in {'mean_pairwise_distance', 'author_group_mean_pairwise_distance'}:
        return float(np.nanmean(np.where(np.eye(len(idx), dtype=bool), np.nan, D[np.ix_(idx, idx)])))
    if metric_name in {'nearest_neighbor_distance', 'author_group_nearest_neighbor_distance'}:
        sub = np.asarray(D[np.ix_(idx, idx)], dtype=float).copy()
        np.fill_diagonal(sub, np.inf)
        return float(np.mean(np.min(sub, axis=1)))
    if metric_name == 'global_nn_distance':
        return _global_knn_group_mean(D, idx, k=1)
    if metric_name == 'global_mean_5nn_distance':
        return _global_knn_group_mean(D, idx, k=5)
    if metric_name == 'mst_dispersion':
        return mst_dispersion(D[np.ix_(idx, idx)])
    if metric_name == 'grid_entropy':
        return grid_entropy(coords, idx, coord_range=coord_range) if coords is not None else np.nan
    raise ValueError(f'Unsupported matrix-derived metric: {metric_name}')


def permutation_pairwise_group_metric(D, idx_a, idx_b, *, metric_name, coords=None, coord_range=None, n_perm=10000, seed=42):
    observed = float(_matrix_group_stat(D, idx_a, metric_name, coords=coords, coord_range=coord_range) - _matrix_group_stat(D, idx_b, metric_name, coords=coords, coord_range=coord_range))
    combined = np.concatenate([idx_a, idx_b])
    n_a = len(idx_a)
    rng = np.random.default_rng(seed)
    null = np.empty(n_perm)
    for i in range(n_perm):
        perm = rng.permutation(combined)
        null[i] = _matrix_group_stat(D, perm[:n_a], metric_name, coords=coords, coord_range=coord_range) - _matrix_group_stat(D, perm[n_a:], metric_name, coords=coords, coord_range=coord_range)
    p = float((np.sum(np.abs(null) >= abs(observed)) + 1) / (n_perm + 1))
    return observed, p


def all_group_matrix_tests(D, group_idx, *, groups=None, metrics=None, coords=None, coord_range=None, condition=None, n_perm=10000, family='proposal_space_matrix_diversity'):
    rows = []
    groups = groups or GROUP_ORDER
    metrics = metrics or (PRIMARY_DIVERSITY_METRICS + ['mst_dispersion'])
    groups = [g for g in groups if g in group_idx and len(group_idx[g]) >= 2]
    for metric in metrics:
        for g1, g2 in combinations(groups, 2):
            effect, p = permutation_pairwise_group_metric(D, group_idx[g1], group_idx[g2], metric_name=metric, coords=coords, coord_range=coord_range, n_perm=n_perm, seed=BOOT_SEED)
            rows.append({
                'condition': condition,
                'analysis_family': family,
                'metric': metric,
                'group_1': g1,
                'group_2': g2,
                'n_1': int(len(group_idx[g1])),
                'n_2': int(len(group_idx[g2])),
                'effect_1_minus_2': effect,
                'permutation_p_value': p,
                'p_value': p,
                'test': 'label permutation',
                'p_adjustment': 'Holm',
                'n_perm': n_perm,
            })
    out = pd.DataFrame(rows)
    if len(out):
        out['q_value_holm'] = out.groupby('metric')['permutation_p_value'].transform(lambda s: holm_q(s).to_numpy())
        out['significance'] = out['q_value_holm'].map(p_stars)
    return out


def matrix_omnibus_group_test(D, group_idx, *, metric_name, groups=None, coords=None, coord_range=None, condition=None, family=None, n_perm=10000):
    groups = groups or GROUP_ORDER
    groups = [g for g in groups if g in group_idx and len(group_idx[g]) >= 2]
    if len(groups) < 2:
        return {'condition': condition, 'analysis_family': family, 'metric': metric_name, 'test': 'label permutation range', 'p_value': np.nan}
    sizes = [len(group_idx[g]) for g in groups]
    observed_stats = np.array([_matrix_group_stat(D, group_idx[g], metric_name, coords=coords, coord_range=coord_range) for g in groups], dtype=float)
    observed = float(np.nanmax(observed_stats) - np.nanmin(observed_stats))
    combined = np.concatenate([group_idx[g] for g in groups])
    rng = np.random.default_rng(BOOT_SEED)
    null = np.empty(n_perm)
    for i in range(n_perm):
        perm = rng.permutation(combined)
        start = 0
        perm_stats = []
        for size in sizes:
            idx = perm[start:start + size]
            start += size
            perm_stats.append(_matrix_group_stat(D, idx, metric_name, coords=coords, coord_range=coord_range))
        perm_stats = np.array(perm_stats, dtype=float)
        null[i] = np.nanmax(perm_stats) - np.nanmin(perm_stats)
    p = float((np.sum(null >= observed) + 1) / (n_perm + 1))
    return {
        'condition': condition,
        'analysis_family': family,
        'metric': metric_name,
        'test': 'label permutation range',
        'statistic': observed,
        'p_value': p,
        'permutation_p_value': p,
        'n_groups': len(groups),
        'n_perm': n_perm,
        'p_adjustment': 'Holm',
    }


def nn_source_composition_permutation_test(nn_df, source_labels, *, groups=None, condition=None, n_perm=10000, seed=42):
    groups = groups or GROUP_ORDER
    source_labels = np.asarray(source_labels, dtype=object)
    nn_idx = nn_df['global_nn_index'].astype(int).to_numpy()
    row_labels = source_labels
    nn_labels = source_labels[nn_idx]
    tab = pd.crosstab(pd.Categorical(row_labels, categories=groups), pd.Categorical(nn_labels, categories=groups), dropna=False)
    chi2, p_chi, dof, _ = chi2_contingency(tab)
    n = tab.to_numpy().sum()
    cramers_v = float(np.sqrt(chi2 / (n * max(1, min(tab.shape) - 1)))) if n else np.nan
    rng = np.random.default_rng(seed)
    null = np.empty(n_perm)
    for i in range(n_perm):
        perm_labels = rng.permutation(source_labels)
        perm_tab = pd.crosstab(pd.Categorical(perm_labels, categories=groups), pd.Categorical(perm_labels[nn_idx], categories=groups), dropna=False)
        perm_chi2, _, _, _ = chi2_contingency(perm_tab)
        null[i] = perm_chi2
    perm_p = float((np.sum(null >= chi2) + 1) / (n_perm + 1))
    return pd.DataFrame([{
        'condition': condition,
        'analysis_family': 'global_nearest_neighbor_source_composition',
        'test': 'label permutation chi-square',
        'chi2': float(chi2),
        'asymptotic_p_value_reference_only': float(p_chi),
        'permutation_p_value': perm_p,
        'p_value': perm_p,
        'dof': int(dof),
        'cramers_v': cramers_v,
        'n_perm': n_perm,
    }])


def compute_author_group_local_metrics(master_df, D):
    rows = []
    for group in GROUP_ORDER:
        idx = master_df.index[master_df['source_group'] == group].to_numpy()
        if len(idx) == 0:
            continue
        sub = np.asarray(D[np.ix_(idx, idx)], dtype=float).copy()
        np.fill_diagonal(sub, np.nan)
        nn = np.asarray(D[np.ix_(idx, idx)], dtype=float).copy()
        np.fill_diagonal(nn, np.inf)
        for pos, global_idx in enumerate(idx):
            rows.append({
                'proposal_uid': master_df.iloc[global_idx]['proposal_uid'],
                'author_group_mean_pairwise_distance': float(np.nanmean(sub[pos])),
                'author_group_nearest_neighbor_distance': float(np.min(nn[pos])),
            })
    return pd.DataFrame(rows)


def compute_global_nn_metrics(master_df, D, k=5):
    arr = np.asarray(D, dtype=float).copy()
    np.fill_diagonal(arr, np.inf)
    nn_idx = arr.argmin(axis=1)
    sorted_idx = np.argsort(arr, axis=1)[:, :k]
    out = master_df[['proposal_uid', 'source_type', 'source_group', 'title']].copy()
    out['global_nn_index'] = nn_idx
    out['global_nn_proposal_uid'] = master_df.iloc[nn_idx]['proposal_uid'].to_numpy()
    out['global_nn_source_type'] = master_df.iloc[nn_idx]['source_type'].to_numpy()
    out['global_nn_source_group'] = master_df.iloc[nn_idx]['source_group'].to_numpy()
    out['global_nn_distance'] = arr[np.arange(len(arr)), nn_idx]
    out['global_mean_5nn_distance'] = np.take_along_axis(arr, sorted_idx, axis=1).mean(axis=1)
    out['global_nn_same_source_type'] = out['source_type'].to_numpy() == out['global_nn_source_type'].to_numpy()
    out['global_nn_same_source_group'] = out['source_group'].to_numpy() == out['global_nn_source_group'].to_numpy()
    out['is_global_nn_outlier_top10'] = out['global_nn_distance'] >= out['global_nn_distance'].quantile(0.90)
    out['is_global_mean5nn_outlier_top10'] = out['global_mean_5nn_distance'] >= out['global_mean_5nn_distance'].quantile(0.90)
    return out


def chamfer_between_groups(D, group_idx):
    rows = []
    for g1, g2 in combinations(GROUP_ORDER, 2):
        val = chamfer_distance(D, group_idx[g1], group_idx[g2])
        rows.append({'group_1': g1, 'group_2': g2, 'chamfer_distance': val})
    return pd.DataFrame(rows)


def parse_mesh_terms(value):
    if isinstance(value, list):
        return [str(v) for v in value]
    if pd.isna(value):
        return []
    text = str(value).strip()
    if not text or text == '[]':
        return []
    try:
        parsed = ast.literal_eval(text)
        if isinstance(parsed, list):
            return [str(v) for v in parsed]
    except Exception:
        pass
    return [x.strip() for x in text.split(';') if x.strip()]


def compute_bertopic_region_metrics(master_df, knn_payload, k=20):
    if not LIT_BERTOPIC_ASSIGNMENTS_PATH.exists():
        print(f'Missing BERTopic assignments: {LIT_BERTOPIC_ASSIGNMENTS_PATH}')
        return pd.DataFrame(), pd.DataFrame()
    bt = pd.read_csv(LIT_BERTOPIC_ASSIGNMENTS_PATH).sort_values('article_idx')
    topic_labels = bt['bertopic_topic'].to_numpy()
    label_map = bt.drop_duplicates('bertopic_topic').set_index('bertopic_topic')['bertopic_label'].to_dict()
    neighbor_idx = np.asarray(knn_payload['neighbor_idx'])[:, :k]
    rows = []
    weight_rows = []
    for i, neigh in enumerate(neighbor_idx):
        topics = topic_labels[neigh]
        counts = Counter(int(t) for t in topics if int(t) != -1)
        total = sum(counts.values())
        if total == 0:
            entropy = np.nan
            max_w = np.nan
            unique = 0
            dominant = -1
        else:
            probs = np.array(list(counts.values()), dtype=float) / total
            entropy = float(-(probs * np.log2(probs)).sum())
            max_w = float(probs.max())
            unique = int(len(counts))
            dominant = int(max(counts, key=counts.get))
        uid = master_df.iloc[i]['proposal_uid']
        for topic, count in counts.items():
            weight_rows.append({'proposal_uid': uid, 'bertopic_topic': topic, 'bertopic_label': label_map.get(topic, str(topic)), 'neighbor_count': count, 'neighbor_weight': count / total if total else 0})
        rows.append({'proposal_uid': uid, 'bertopic_region_entropy': entropy, 'bertopic_max_region_weight': max_w, 'bertopic_unique_regions': unique, 'bertopic_dominant_region': dominant, 'bertopic_dominant_label': label_map.get(dominant, str(dominant))})
    return pd.DataFrame(rows), pd.DataFrame(weight_rows)


def compute_mesh_metrics(master_df, knn_payload, k=20):
    if not LIT_ARTICLE_INDEX_PATH.exists():
        print(f'Missing literature article index: {LIT_ARTICLE_INDEX_PATH}')
        return pd.DataFrame(), pd.DataFrame()
    article_df = pd.read_csv(LIT_ARTICLE_INDEX_PATH).sort_values('article_idx')
    mesh_sets = [set(parse_mesh_terms(v)) for v in article_df['mesh_terms']]
    neighbor_idx = np.asarray(knn_payload['neighbor_idx'])[:, :k]
    rows = []
    term_rows = []
    for i, neigh in enumerate(neighbor_idx):
        terms = set()
        for j in neigh:
            terms.update(mesh_sets[int(j)])
        uid = master_df.iloc[i]['proposal_uid']
        rows.append({'proposal_uid': uid, 'mesh_unique_terms_k20': len(terms)})
        for term in sorted(terms):
            term_rows.append({'proposal_uid': uid, 'mesh_term': term})
    return pd.DataFrame(rows), pd.DataFrame(term_rows)


def plot_chamfer_heatmap(chamfer_df, out_path, title, n_map=None):
    mat = pd.DataFrame(np.nan, index=GROUP_ORDER, columns=GROUP_ORDER)
    for _, r in chamfer_df.iterrows():
        mat.loc[r['group_1'], r['group_2']] = r['chamfer_distance']
        mat.loc[r['group_2'], r['group_1']] = r['chamfer_distance']
    np.fill_diagonal(mat.values, 0)
    mat_for_plot = relabel_index_with_n(mat, n_map)
    if n_map is not None:
        mat_for_plot.columns = [label_with_n(c, n_map) for c in mat.columns]
    plot_heatmap(mat_for_plot, title=title, out_path=out_path, fmt='.3f', cmap='mako')


def load_literature_umap_assets():
    if not (LIT_UMAP2D_PATH.exists() and LIT_BERTOPIC_ASSIGNMENTS_PATH.exists() and LIT_UMAP_REDUCER_PATH.exists()):
        return None, None, None, None
    lit_2d = np.load(LIT_UMAP2D_PATH)
    bt = pd.read_csv(LIT_BERTOPIC_ASSIGNMENTS_PATH).sort_values('article_idx')
    topic_info = pd.read_csv(LIT_BERTOPIC_TOPIC_INFO_PATH) if LIT_BERTOPIC_TOPIC_INFO_PATH.exists() else pd.DataFrame()
    with open(LIT_UMAP_REDUCER_PATH, 'rb') as f:
        reducer = pickle.load(f)
    return lit_2d, bt, topic_info, reducer


def plot_literature_umap_overlay(condition, analysis, master_df, figures_dir, *, zoom=False, n_map=None):
    assets = load_literature_umap_assets()
    lit_2d, bt, topic_info, reducer = assets
    if lit_2d is None:
        print('Literature UMAP/BERTopic assets missing; skipping literature overlay.')
        return
    X_abs = np.asarray(analysis.abstract_embeddings['embeddings'], dtype=np.float32)
    prop_2d = reducer.transform(X_abs)
    plot_df = master_df.copy()
    plot_df['lit_umap_x'] = prop_2d[:, 0]
    plot_df['lit_umap_y'] = prop_2d[:, 1]
    if zoom:
        xpad = (plot_df['lit_umap_x'].max() - plot_df['lit_umap_x'].min()) * 0.25 + 1e-6
        ypad = (plot_df['lit_umap_y'].max() - plot_df['lit_umap_y'].min()) * 0.25 + 1e-6
        xlim = (plot_df['lit_umap_x'].min() - xpad, plot_df['lit_umap_x'].max() + xpad)
        ylim = (plot_df['lit_umap_y'].min() - ypad, plot_df['lit_umap_y'].max() + ypad)
        mask = (lit_2d[:, 0] >= xlim[0]) & (lit_2d[:, 0] <= xlim[1]) & (lit_2d[:, 1] >= ylim[0]) & (lit_2d[:, 1] <= ylim[1])
        suffix = 'zoomed'
    else:
        mask = np.ones(len(lit_2d), dtype=bool)
        xlim = ylim = None
        suffix = 'full'

    n_map = n_map or group_n_map(plot_df, 'source_group', GROUP_ORDER)
    fig, ax = plt.subplots(figsize=(10.5, 8.2))
    ax.scatter(lit_2d[mask, 0], lit_2d[mask, 1], c=bt.loc[mask, 'bertopic_topic'], cmap='tab20', s=3, alpha=0.16, linewidths=0, rasterized=True)
    for g in GROUP_ORDER:
        sub = plot_df[plot_df['source_group'] == g]
        ax.scatter(sub['lit_umap_x'], sub['lit_umap_y'], s=58, c=GROUP_PALETTE.get(g, '#808080'), label=label_with_n(g, n_map), edgecolor='white', linewidth=0.7, alpha=0.95)
        funded = sub[sub['is_funded_human'].map(truthy_flag)] if 'is_funded_human' in sub else sub.iloc[0:0]
        if len(funded):
            ax.scatter(funded['lit_umap_x'], funded['lit_umap_y'], s=110, facecolors='none', edgecolors='#FF00FF', linewidths=1.6, label='Accepted/Funded Human ring')
        top_ai_col = 'is_top5_AI-ranked' if 'is_top5_AI-ranked' in sub else ('is_top5_AI_ranked' if 'is_top5_AI_ranked' in sub else None)
        top = sub[sub[top_ai_col].map(truthy_flag)] if top_ai_col else sub.iloc[0:0]
        if len(top):
            ax.scatter(top['lit_umap_x'], top['lit_umap_y'], s=135, facecolors='none', edgecolors='#FF8C00', linewidths=1.9, label='Top AI-ranked Human ring')
    if topic_info is not None and len(topic_info):
        label_rows = []
        topics = bt['bertopic_topic'].to_numpy()
        for _, row in topic_info.iterrows():
            topic = row.get('Topic')
            tmask = mask & (topics == topic)
            n_visible = int(tmask.sum())
            if n_visible < 80:
                continue
            label_rows.append({
                'topic': topic,
                'n_visible': n_visible,
                'x': float(np.nanmedian(lit_2d[tmask, 0])),
                'y': float(np.nanmedian(lit_2d[tmask, 1])),
                'label': str(row.get('display_label', topic))[:30],
            })
        for row in sorted(label_rows, key=lambda r: r['n_visible'], reverse=True)[:8]:
            ax.annotate(
                row['label'],
                xy=(row['x'], row['y']),
                xytext=(6, 6),
                textcoords='offset points',
                fontsize=6.5,
                ha='left',
                va='bottom',
                bbox={'facecolor': 'white', 'alpha': 0.72, 'edgecolor': 'none', 'pad': 1.2},
                arrowprops={'arrowstyle': '-', 'color': '#666666', 'linewidth': 0.45, 'alpha': 0.65},
                clip_on=True,
            )
    if xlim:
        ax.set_xlim(*xlim)
        ax.set_ylim(*ylim)
    ax.set_title(f'{condition}: literature-anchored BERTopic map ({suffix})')
    ax.set_xlabel('Literature UMAP Dim 1')
    ax.set_ylabel('Literature UMAP Dim 2')
    handles, labels = ax.get_legend_handles_labels()
    by_label = dict(zip(labels, handles))
    ax.legend(by_label.values(), by_label.keys(), loc='upper left', bbox_to_anchor=(1.02, 1.0), frameon=False, fontsize=9)
    save_show(fig, figures_dir / f'literature_umap_bertopic_overlay_{suffix}.png')


def plot_outlier_prevalence(df, flag_cols, out_path, title, n_map=None):
    rows = []
    for flag in flag_cols:
        if flag not in df.columns:
            continue
        for group, sub in df.groupby('source_group'):
            rows.append({'flag': flag, 'source_group': group, 'prevalence': float(sub[flag].mean()), 'count': int(sub[flag].sum()), 'n': len(sub)})
    plot_df = pd.DataFrame(rows)
    if plot_df.empty:
        return
    fig, ax = plt.subplots(figsize=(10, 5))
    sns.barplot(data=plot_df, x='flag', y='prevalence', hue='source_group', hue_order=GROUP_ORDER, palette=GROUP_PALETTE, ax=ax)
    ax.set_title(title)
    ax.set_ylabel('Flag prevalence')
    ax.set_xlabel('Outlier flag')
    ax.tick_params(axis='x', rotation=20)
    n_map = n_map or group_n_map(df, 'source_group')
    relabel_legend_with_n(ax, n_map)
    save_show(fig, out_path)



def _bootstrap_summary_row(condition, metric, family, human_value, ai_boot_values, *, n_human, n_ai_pool, n_boot):
    ai_boot_values = np.asarray(pd.Series(ai_boot_values).dropna(), dtype=float)
    if ai_boot_values.size == 0 or pd.isna(human_value):
        return {
            'condition': condition,
            'analysis_family': family,
            'metric': metric,
            'test': 'same-size AI bootstrap',
            'human_n': n_human,
            'ai_subsample_n': n_human,
            'ai_pool_n': n_ai_pool,
            'n_boot': n_boot,
            'human_value': human_value,
            'ai_boot_mean': np.nan,
            'ai_boot_ci_low': np.nan,
            'ai_boot_ci_high': np.nan,
            'effect_human_minus_ai_boot_mean': np.nan,
            'effect_ci_low': np.nan,
            'effect_ci_high': np.nan,
            'bootstrap_p_value_directional': np.nan,
        }
    effects = float(human_value) - ai_boot_values
    p_lower = float(np.mean(effects <= 0))
    p_upper = float(np.mean(effects >= 0))
    p_boot = min(1.0, 2.0 * min(p_lower, p_upper))
    return {
        'condition': condition,
        'analysis_family': family,
        'metric': metric,
        'test': 'same-size AI bootstrap',
        'human_n': n_human,
        'ai_subsample_n': n_human,
        'ai_pool_n': n_ai_pool,
        'n_boot': int(ai_boot_values.size),
        'human_value': float(human_value),
        'ai_boot_mean': float(np.mean(ai_boot_values)),
        'ai_boot_ci_low': float(np.percentile(ai_boot_values, 2.5)),
        'ai_boot_ci_high': float(np.percentile(ai_boot_values, 97.5)),
        'effect_human_minus_ai_boot_mean': float(np.mean(effects)),
        'effect_ci_low': float(np.percentile(effects, 2.5)),
        'effect_ci_high': float(np.percentile(effects, 97.5)),
        'bootstrap_p_value_directional': p_boot,
    }


def pooled_bootstrap_metric_values(master_df, D, human_idx, boot_ai_idx_samples, metric, *, coords=None, coord_range=None):
    if metric in MATRIX_DERIVED_METRICS or metric in {'mst_dispersion', 'grid_entropy'}:
        human_value = _matrix_group_stat(D, human_idx, metric, coords=coords, coord_range=coord_range)
        ai_boot_values = np.array([_matrix_group_stat(D, sample, metric, coords=coords, coord_range=coord_range) for sample in boot_ai_idx_samples], dtype=float)
    else:
        values = pd.to_numeric(master_df[metric], errors='coerce').to_numpy(dtype=float)
        human_value = float(np.nanmean(values[human_idx]))
        ai_boot_values = np.array([np.nanmean(values[sample]) for sample in boot_ai_idx_samples], dtype=float)
    return human_value, ai_boot_values


def plot_pooled_bootstrap_distribution(ai_boot_values, human_value, *, condition, metric, family, out_path, n_human, n_ai_pool):
    ai_boot_values = np.asarray(pd.Series(ai_boot_values).dropna(), dtype=float)
    if ai_boot_values.size == 0 or pd.isna(human_value):
        print(f'No bootstrap data to plot for {metric}')
        return
    ci_low, ci_high = np.percentile(ai_boot_values, [2.5, 97.5])
    metric_label = f'{metric} (fixed condition-level UMAP range)' if metric == 'grid_entropy' else metric
    fig, ax = plt.subplots(figsize=(8.5, 4.8))
    sns.histplot(ai_boot_values, kde=True, color=GROUP_PALETTE.get('All AI', '#4A90E2'), alpha=0.45, ax=ax)
    ax.axvspan(ci_low, ci_high, color=GROUP_PALETTE.get('All AI', '#4A90E2'), alpha=0.16, label='AI bootstrap 95% interval')
    ax.axvline(float(human_value), color=GROUP_PALETTE['Human'], linewidth=2.4, label=f'Human n={n_human}')
    ax.axvline(float(np.mean(ai_boot_values)), color=GROUP_PALETTE.get('All AI', '#4A90E2'), linewidth=2.0, linestyle='--', label=f'AI bootstrap mean')
    ax.set_title(f'{condition}: {metric_label}\nHuman n={n_human} vs All AI bootstrap n={n_human} from {n_ai_pool}; B={len(ai_boot_values)}')
    ax.set_xlabel(metric_label)
    ax.set_ylabel('Bootstrap samples')
    ax.legend(loc='upper left', bbox_to_anchor=(1.02, 1.0), frameon=False, fontsize=9)
    save_show(fig, out_path)


def pooled_bootstrap_metric_suite(master_df, D, human_idx, ai_idx, boot_ai_idx_samples, metrics, *, coords=None, coord_range=None, condition=None, family=None, out_dir=None):
    rows = []
    long_rows = []
    for metric in metrics:
        if metric not in master_df.columns and metric not in MATRIX_DERIVED_METRICS and metric not in {'mst_dispersion', 'grid_entropy'}:
            continue
        human_value, ai_boot_values = pooled_bootstrap_metric_values(master_df, D, human_idx, boot_ai_idx_samples, metric, coords=coords, coord_range=coord_range)
        rows.append(_bootstrap_summary_row(condition, metric, family, human_value, ai_boot_values, n_human=len(human_idx), n_ai_pool=len(ai_idx), n_boot=len(ai_boot_values)))
        for sample_idx, value in enumerate(ai_boot_values):
            long_rows.append({'condition': condition, 'analysis_family': family, 'metric': metric, 'sample_index': sample_idx, 'ai_boot_value': float(value), 'human_value': float(human_value), 'effect_human_minus_ai': float(human_value - value)})
        if out_dir is not None:
            plot_pooled_bootstrap_distribution(ai_boot_values, human_value, condition=condition, metric=metric, family=family, out_path=out_dir / f'{metric}_bootstrap_distribution.png', n_human=len(human_idx), n_ai_pool=len(ai_idx))
    return pd.DataFrame(rows), pd.DataFrame(long_rows)


def bootstrap_mesh_union_terms(mesh_terms_df, master_df, human_idx, ai_idx, boot_ai_idx_samples, *, condition=None, out_dir=None):
    if mesh_terms_df.empty:
        return pd.DataFrame(), pd.DataFrame()
    uid_by_pos = master_df['proposal_uid'].astype(str).to_numpy()
    terms_by_uid = mesh_terms_df.groupby('proposal_uid')['mesh_term'].apply(lambda s: set(s.dropna().astype(str))).to_dict()
    def union_count(idx):
        terms = set()
        for uid in uid_by_pos[idx]:
            terms.update(terms_by_uid.get(str(uid), set()))
        return len(terms)
    human_value = float(union_count(human_idx))
    ai_boot_values = np.array([union_count(sample) for sample in boot_ai_idx_samples], dtype=float)
    row = _bootstrap_summary_row(condition, 'mesh_group_union_terms', 'pooled_human_vs_allai_mesh_union', human_value, ai_boot_values, n_human=len(human_idx), n_ai_pool=len(ai_idx), n_boot=len(ai_boot_values))
    long_df = pd.DataFrame({'condition': condition, 'metric': 'mesh_group_union_terms', 'sample_index': np.arange(len(ai_boot_values)), 'ai_boot_value': ai_boot_values, 'human_value': human_value, 'effect_human_minus_ai': human_value - ai_boot_values})
    if out_dir is not None:
        plot_pooled_bootstrap_distribution(ai_boot_values, human_value, condition=condition, metric='mesh_group_union_terms', family='pooled_human_vs_allai_mesh_union', out_path=out_dir / 'mesh_group_union_terms_bootstrap_distribution.png', n_human=len(human_idx), n_ai_pool=len(ai_idx))
    return pd.DataFrame([row]), long_df


def bootstrap_bertopic_region_weights(bertopic_weights_df, master_df, human_idx, ai_idx, boot_ai_idx_samples, *, condition=None, out_dir=None, top_n=10):
    if bertopic_weights_df.empty:
        return pd.DataFrame(), pd.DataFrame()
    uid_by_pos = master_df['proposal_uid'].astype(str).to_numpy()
    wide = bertopic_weights_df.pivot_table(index='proposal_uid', columns='bertopic_label', values='neighbor_weight', aggfunc='sum', fill_value=0)
    wide = wide.reindex(uid_by_pos, fill_value=0)
    human_weights = wide.iloc[human_idx].mean(axis=0)
    ai_boot = []
    for sample_idx, sample in enumerate(boot_ai_idx_samples):
        vals = wide.iloc[sample].mean(axis=0)
        for label, value in vals.items():
            ai_boot.append({'sample_index': sample_idx, 'bertopic_label': label, 'ai_boot_weight': float(value)})
    ai_boot_df = pd.DataFrame(ai_boot)
    top_labels = (human_weights.add(ai_boot_df.groupby('bertopic_label')['ai_boot_weight'].mean(), fill_value=0).sort_values(ascending=False).head(top_n).index.tolist())
    rows = []
    for label in top_labels:
        vals = ai_boot_df.loc[ai_boot_df['bertopic_label'] == label, 'ai_boot_weight'].to_numpy(dtype=float)
        rows.append({
            'condition': condition,
            'bertopic_label': label,
            'human_n': len(human_idx),
            'ai_subsample_n': len(human_idx),
            'ai_pool_n': len(ai_idx),
            'n_boot': len(boot_ai_idx_samples),
            'human_weight': float(human_weights.get(label, 0.0)),
            'ai_boot_mean_weight': float(np.mean(vals)),
            'ai_boot_ci_low': float(np.percentile(vals, 2.5)),
            'ai_boot_ci_high': float(np.percentile(vals, 97.5)),
        })
    ci_df = pd.DataFrame(rows)
    if out_dir is not None and len(ci_df):
        fig, ax = plt.subplots(figsize=(9, max(4.8, 0.42 * len(ci_df))))
        y = np.arange(len(ci_df))
        ax.errorbar(ci_df['ai_boot_mean_weight'], y, xerr=[ci_df['ai_boot_mean_weight'] - ci_df['ai_boot_ci_low'], ci_df['ai_boot_ci_high'] - ci_df['ai_boot_mean_weight']], fmt='o', color=GROUP_PALETTE.get('All AI', '#4A90E2'), label=f'All AI bootstrap n={len(human_idx)} from {len(ai_idx)}')
        ax.scatter(ci_df['human_weight'], y, marker='D', color=GROUP_PALETTE['Human'], label=f'Human n={len(human_idx)}', zorder=5)
        ax.set_yticks(y)
        ax.set_yticklabels(ci_df['bertopic_label'])
        ax.invert_yaxis()
        ax.set_xlabel('Mean nearest-literature region weight')
        ax.set_title(f'{condition}: BERTopic region coverage\nHuman n={len(human_idx)} vs All AI bootstrap n={len(human_idx)} from {len(ai_idx)}; B={len(boot_ai_idx_samples)}')
        ax.legend(loc='upper left', bbox_to_anchor=(1.02, 1.0), frameon=False, fontsize=9)
        save_show(fig, out_dir / 'bertopic_region_coverage_bootstrap_ci.png')
    return ci_df, ai_boot_df
```

```python

condition_results = {}
cross_condition_pooled_rows = []
cross_condition_per_model_rows = []
cross_condition_metric_frames = []
cross_condition_facet_test_frames = []
cross_condition_facet_gradient_frames = []
cross_condition_facet_curve_frames = []
analysis_output_rows = []

for condition in CONDITIONS_TO_RUN:
    print(f'\n=== 5b rephrased proposal comparison: {condition} ===')
    analysis = load_condition_analysis_inputs(PROJECT_ROOT, condition, text_version=TEXT_VERSION)
    group_idx = get_group_indices(analysis.proposal_master)
    human_idx = group_idx['Human']
    ai_idx = group_idx['All AI']

    if len(human_idx) != 23:
        raise RuntimeError(f'{condition}: expected 23 Human proposals, found {len(human_idx)}')
    if len(ai_idx) != 69:
        raise RuntimeError(f'{condition}: expected 69 pooled AI proposals, found {len(ai_idx)}')
    for model_label in MODEL_ORDER:
        if len(group_idx.get(model_label, [])) != 23:
            raise RuntimeError(f'{condition}: expected 23 proposals for {model_label}, found {len(group_idx.get(model_label, []))}')

    author_group_n = {g: int(len(group_idx.get(g, []))) for g in GROUP_ORDER}
    pooled_group_n = {'Human': int(len(human_idx)), 'All AI': int(len(ai_idx))}
    condition_umap_range = fixed_coordinate_range(analysis.umap2d)

    tables_root = PROJECT_ROOT / 'results' / 'tables' / condition / 'proposals' / TEXT_VERSION
    figures_root = PROJECT_ROOT / 'results' / 'figures' / condition / 'proposals' / TEXT_VERSION
    per_model_tables_dir = tables_root / 'per_model'
    per_model_figures_dir = figures_root / 'per_model'
    pooled_tables_dir = tables_root / 'all_ai'
    pooled_figures_dir = figures_root / 'all_ai'
    shared_cache_dir = tables_root / 'shared_cache'
    facet_tables_dir = tables_root / 'facet'
    facet_figures_dir = figures_root / 'facet'
    for p in [per_model_tables_dir, per_model_figures_dir, pooled_tables_dir, pooled_figures_dir, shared_cache_dir, facet_tables_dir, facet_figures_dir]:
        p.mkdir(parents=True, exist_ok=True)

    bootstrap_path_original = PROJECT_ROOT / 'results' / 'tables' / condition / 'proposals' / 'original' / 'shared_cache' / f'bootstrap_ai_idx_samples_n{N_SUBSAMPLE}_seed{BOOT_SEED}.npy'
    bootstrap_path_rephrased = shared_cache_dir / f'bootstrap_ai_idx_samples_n{N_SUBSAMPLE}_seed{BOOT_SEED}.npy'
    original_master_path = PROJECT_ROOT / 'data' / 'prepared' / condition / 'proposals' / 'original' / 'proposal_master.csv'
    if bootstrap_path_original.exists() and original_master_path.exists() and analysis.proposal_master['proposal_uid'].astype(str).tolist() == pd.read_csv(original_master_path)['proposal_uid'].astype(str).tolist():
        bootstrap_ai_idx_samples = np.load(bootstrap_path_original)
        np.save(bootstrap_path_rephrased, bootstrap_ai_idx_samples)
    else:
        bootstrap_ai_idx_samples = generate_or_load_bootstrap_samples(
            ai_idx,
            output_path=bootstrap_path_rephrased,
            n_boot=N_BOOT,
            n_subsample=N_SUBSAMPLE,
            seed=BOOT_SEED,
        )


    # Facet Diversity (M1-M5): full-D embedding metrics, equal-n inference, and tidy exports.
    facet_tests_df, facet_gradient_df, facet_curves_df = build_proposal_facet_outputs(
        analysis,
        text_branch=TEXT_VERSION,
        bootstrap_ai_idx_samples=bootstrap_ai_idx_samples,
        n_perm=N_PERM,
        n_boot=N_BOOT,
        seed=BOOT_SEED,
    )
    write_facet_outputs(
        facet_tables_dir,
        facet_figures_dir,
        facet_tests_df,
        facet_gradient_df,
        facet_curves_df,
    )
    cross_condition_facet_test_frames.append(facet_tests_df)
    cross_condition_facet_gradient_frames.append(facet_gradient_df)
    cross_condition_facet_curve_frames.append(facet_curves_df)

    lit_self_knn_path = shared_cache_dir / 'lit_knn_distances_50.npy'
    lit_knn_distances_50 = compute_literature_self_knn_cache(
        LITERATURE_EMBEDDINGS_PATH,
        output_path=lit_self_knn_path,
        k=50,
    )
    lit_mean_knn_10 = lit_knn_distances_50[:, :10].mean(axis=1)

    proposal_metrics_master = build_proposal_metrics_master(analysis, lit_self_knn_mean10=lit_mean_knn_10)
    author_local_df = compute_author_group_local_metrics(analysis.proposal_master, analysis.pairwise_full)
    proposal_metrics_master = proposal_metrics_master.merge(author_local_df, on='proposal_uid', how='left', validate='one_to_one')

 
    # Matrix-derived diversity tests for every author group pair, not only Human vs model.
    matrix_pairwise_tests = all_group_matrix_tests(analysis.pairwise_full, group_idx, metrics=list(dict.fromkeys([*MATRIX_DERIVED_METRICS, *PRIMARY_DIVERSITY_METRICS, 'mst_dispersion'])), coords=analysis.umap2d, coord_range=condition_umap_range, condition=condition, n_perm=N_PERM)
    matrix_pairwise_tests.to_csv(per_model_tables_dir / 'matrix_diversity_pairwise_tests_all_author_groups.csv', index=False)

    per_model_rows = []
    for model_label in MODEL_ORDER:
        model_idx = group_idx[model_label]
        metrics_df = compare_human_vs_group_metrics(
            analysis.pairwise_full,
            human_idx,
            model_idx,
            coords=analysis.umap2d,
            coord_range=condition_umap_range,
            metric_names=PRIMARY_DIVERSITY_METRICS + ['mst_dispersion'],
            n_perm=N_PERM,
            seed=BOOT_SEED,
        )
        metrics_df['condition'] = condition
        metrics_df['comparison_group'] = model_label
        per_model_rows.append(metrics_df)
        for _, row in metrics_df.iterrows():
            cross_condition_per_model_rows.append({
                'condition': condition,
                'comparison_group': model_label,
                'metric': row['metric'],
                'effect_human_minus_group': row['effect_human_minus_group'],
                'permutation_p_value': row['permutation_p_value'],
                'analysis_family': 'proposal_space_diversity',
            })
    per_model_summary_df = pd.concat(per_model_rows, ignore_index=True)
    per_model_summary_df.to_csv(per_model_tables_dir / 'diversity_summary_human_vs_per_model.csv', index=False)

    pooled_rows = []
    bootstrap_frames = []
    for metric_name in PRIMARY_DIVERSITY_METRICS + ['mst_dispersion']:
        boot_df = pooled_bootstrap_metric_distribution(
            analysis.pairwise_full,
            human_idx,
            bootstrap_ai_idx_samples,
            metric_name=metric_name,
            coords=analysis.umap2d,
            coord_range=condition_umap_range,
        )
        pooled_metric_df = compare_human_vs_group_metrics(
            analysis.pairwise_full,
            human_idx,
            ai_idx,
            coords=analysis.umap2d,
            coord_range=condition_umap_range,
            metric_names=[metric_name],
            n_perm=N_PERM,
            seed=BOOT_SEED,
        )
        pooled_row = pooled_metric_df.iloc[0].to_dict()
        pooled_row.update({
            'condition': condition,
            'human_value': float(boot_df['human_value'].iloc[0]),
            'ai_boot_mean': float(boot_df['ai_value'].mean()),
            'ai_boot_sd': float(boot_df['ai_value'].std()),
        })
        pooled_rows.append(pooled_row)
        bootstrap_frames.append(boot_df.assign(condition=condition))
    pooled_summary_df = pd.DataFrame(pooled_rows)
    pooled_summary_df.to_csv(pooled_tables_dir / 'diversity_summary_human_vs_allai_unequal_n_reference_only.csv', index=False)
    pd.concat(bootstrap_frames, ignore_index=True).to_csv(pooled_tables_dir / 'proposal_metrics_summary_human_vs_allai.csv', index=False)

    
    # BERTopic literature-region coverage from high-dimensional nearest literature neighbors.
    bertopic_metrics_df, bertopic_weights_df = compute_bertopic_region_metrics(analysis.proposal_master, analysis.proposal_to_literature_knn, k=20)
    if len(bertopic_metrics_df):
        bertopic_metrics_df = bertopic_metrics_df.merge(analysis.proposal_master[['proposal_uid', 'source_group', 'source_type', 'title']], on='proposal_uid', how='left')
        bertopic_metrics_df.to_csv(per_model_tables_dir / 'bertopic_region_coverage_per_proposal.csv', index=False)
        bertopic_weights_df.to_csv(per_model_tables_dir / 'bertopic_region_neighbor_weights_long.csv', index=False)
        proposal_metrics_master = proposal_metrics_master.merge(bertopic_metrics_df.drop(columns=['source_group', 'source_type', 'title']), on='proposal_uid', how='left', validate='one_to_one')
        bt_tests = pd.concat([pairwise_mw_tests(proposal_metrics_master, m, condition=condition, family='bertopic_region_coverage') for m in BERTOPIC_METRICS], ignore_index=True)
        bt_tests['facet'] = 'coverage'
        bt_tests['coverage_metric_family'] = 'coverage_bertopic_region'
        bt_tests.to_csv(per_model_tables_dir / 'bertopic_region_coverage_pairwise_tests.csv', index=False)
        group_region = bertopic_weights_df.merge(analysis.proposal_master[['proposal_uid', 'source_group']], on='proposal_uid', how='left').groupby(['source_group', 'bertopic_label'], as_index=False)['neighbor_weight'].mean()
        top_labels = group_region.groupby('bertopic_label')['neighbor_weight'].sum().sort_values(ascending=False).head(10).index
        plot_weighted_stacked(group_region[group_region['bertopic_label'].isin(top_labels)], 'source_group', 'bertopic_label', 'neighbor_weight', title=f'{condition}: BERTopic literature-region coverage', out_path=per_model_figures_dir / 'bertopic_region_coverage_stacked.png', n_map=author_group_n)
        for metric in BERTOPIC_METRICS:
            plot_box_ci_forest(proposal_metrics_master, metric, title=f'{condition}: {metric}', out_path=per_model_figures_dir / f'{metric}_box_forest.png', tests_df=bt_tests, n_map=author_group_n)
        plot_literature_umap_overlay(condition, analysis, proposal_metrics_master, per_model_figures_dir, zoom=False, n_map=author_group_n)
        plot_literature_umap_overlay(condition, analysis, proposal_metrics_master, per_model_figures_dir, zoom=True, n_map=author_group_n)

    # MeSH coverage from nearest literature neighbors.
    mesh_metrics_df, mesh_terms_df = compute_mesh_metrics(analysis.proposal_master, analysis.proposal_to_literature_knn, k=20)
    if len(mesh_metrics_df):
        mesh_metrics_df = mesh_metrics_df.merge(analysis.proposal_master[['proposal_uid', 'source_group', 'source_type', 'title']], on='proposal_uid', how='left')
        mesh_metrics_df.to_csv(per_model_tables_dir / 'mesh_coverage_per_proposal.csv', index=False)
        mesh_terms_df.to_csv(per_model_tables_dir / 'mesh_terms_by_proposal_long.csv', index=False)
        group_mesh_union = mesh_terms_df.merge(analysis.proposal_master[['proposal_uid', 'source_group']], on='proposal_uid', how='left').groupby('source_group')['mesh_term'].nunique().reset_index(name='group_unique_mesh_terms')
        group_mesh_union.to_csv(per_model_tables_dir / 'mesh_coverage_group_union.csv', index=False)
        proposal_metrics_master = proposal_metrics_master.merge(mesh_metrics_df[['proposal_uid', 'mesh_unique_terms_k20']], on='proposal_uid', how='left', validate='one_to_one')
        mesh_tests = pd.concat([pairwise_mw_tests(proposal_metrics_master, m, condition=condition, family='mesh_term_coverage') for m in MESH_METRICS], ignore_index=True)
        mesh_tests['facet'] = 'coverage'
        mesh_tests['coverage_metric_family'] = 'coverage_mesh_terms'
        mesh_tests.to_csv(per_model_tables_dir / 'mesh_coverage_pairwise_tests.csv', index=False)
        plot_box_ci_forest(proposal_metrics_master, 'mesh_unique_terms_k20', title=f'{condition}: MeSH term coverage from nearest literature', out_path=per_model_figures_dir / 'mesh_unique_terms_k20_box_forest.png', tests_df=mesh_tests, n_map=author_group_n)
        fig, ax = plt.subplots(figsize=(8, 5))
        sns.barplot(data=group_mesh_union, x='source_group', y='group_unique_mesh_terms', order=GROUP_ORDER, palette=GROUP_PALETTE, ax=ax)
        ax.set_title(f'{condition}: group-level MeSH union coverage')
        ax.set_xlabel('Author group')
        ax.set_ylabel('Unique MeSH descriptors')
        apply_group_n_xticklabels(ax, GROUP_ORDER, author_group_n, rotation=20)
        save_show(fig, per_model_figures_dir / 'mesh_group_union_coverage.png')

    # Pooled Human-vs-All-AI versions use same-size AI bootstrap samples.
    # Human is fixed at n=23; each All-AI sample is n=23 drawn from the 69 AI proposals.
    pooled_group_order = ['Human', 'All AI']
    pooled_matrix_metrics = list(dict.fromkeys([*PRIMARY_DIVERSITY_METRICS, 'mst_dispersion', *NN_METRICS]))
    pooled_matrix_boot_summary, pooled_matrix_boot_long = pooled_bootstrap_metric_suite(
        proposal_metrics_master,
        analysis.pairwise_full,
        human_idx,
        ai_idx,
        bootstrap_ai_idx_samples,
        pooled_matrix_metrics,
        coords=analysis.umap2d,
        coord_range=condition_umap_range,
        condition=condition,
        family='pooled_human_vs_allai_matrix_same_n_bootstrap',
        out_dir=pooled_figures_dir,
    )
    pooled_matrix_boot_summary.to_csv(pooled_tables_dir / 'pooled_matrix_metric_bootstrap_summaries.csv', index=False)
    pooled_matrix_boot_long.to_csv(pooled_tables_dir / 'pooled_matrix_metric_bootstrap_values_long.csv', index=False)

    pooled_independent_metrics = [m for m in [*NOVELTY_METRICS, *BERTOPIC_METRICS, *MESH_METRICS] if m in proposal_metrics_master.columns]
    pooled_external_boot_summary, pooled_external_boot_long = pooled_bootstrap_metric_suite(
        proposal_metrics_master,
        analysis.pairwise_full,
        human_idx,
        ai_idx,
        bootstrap_ai_idx_samples,
        pooled_independent_metrics,
        coords=analysis.umap2d,
        coord_range=condition_umap_range,
        condition=condition,
        family='pooled_human_vs_allai_external_reference_same_n_bootstrap',
        out_dir=pooled_figures_dir,
    )
    pooled_external_boot_summary.to_csv(pooled_tables_dir / 'pooled_external_metric_bootstrap_summaries.csv', index=False)
    pooled_external_boot_long.to_csv(pooled_tables_dir / 'pooled_external_metric_bootstrap_values_long.csv', index=False)

    mesh_union_boot_summary, mesh_union_boot_long = bootstrap_mesh_union_terms(
        mesh_terms_df,
        analysis.proposal_master,
        human_idx,
        ai_idx,
        bootstrap_ai_idx_samples,
        condition=condition,
        out_dir=pooled_figures_dir,
    )
    if len(mesh_union_boot_summary):
        mesh_union_boot_summary.to_csv(pooled_tables_dir / 'mesh_group_union_bootstrap_summary.csv', index=False)
        mesh_union_boot_long.to_csv(pooled_tables_dir / 'mesh_group_union_bootstrap_values_long.csv', index=False)

    bertopic_region_boot_ci, bertopic_region_boot_long = bootstrap_bertopic_region_weights(
        bertopic_weights_df,
        analysis.proposal_master,
        human_idx,
        ai_idx,
        bootstrap_ai_idx_samples,
        condition=condition,
        out_dir=pooled_figures_dir,
    )
    if len(bertopic_region_boot_ci):
        bertopic_region_boot_ci.to_csv(pooled_tables_dir / 'bertopic_region_coverage_bootstrap_ci.csv', index=False)
        bertopic_region_boot_long.to_csv(pooled_tables_dir / 'bertopic_region_coverage_bootstrap_values_long.csv', index=False)

    pooled_bootstrap_summary_df = pd.concat(
        [x for x in [pooled_matrix_boot_summary, pooled_external_boot_summary, mesh_union_boot_summary] if len(x)],
        ignore_index=True,
        sort=False,
    ) if (len(pooled_matrix_boot_summary) or len(pooled_external_boot_summary) or len(mesh_union_boot_summary)) else pd.DataFrame()
    pooled_bootstrap_summary_df.to_csv(pooled_tables_dir / 'pooled_all_metric_bootstrap_summaries.csv', index=False)
    if len(pooled_bootstrap_summary_df):
        pooled_bootstrap_summary_df[pooled_bootstrap_summary_df['metric'].isin(PRIMARY_DIVERSITY_METRICS + ['mst_dispersion'])].to_csv(pooled_tables_dir / 'diversity_summary_human_vs_allai.csv', index=False)
        for _, row in pooled_bootstrap_summary_df[pooled_bootstrap_summary_df['metric'].isin(PRIMARY_DIVERSITY_METRICS)].iterrows():
            cross_condition_pooled_rows.append({
                'condition': condition,
                'metric': row['metric'],
                'effect_human_minus_ai': row['effect_human_minus_ai_boot_mean'],
                'permutation_p_value': np.nan,
                'bootstrap_p_value_directional': row.get('bootstrap_p_value_directional', np.nan),
                'analysis_family': 'proposal_space_diversity_same_n_bootstrap',
            })

    topic_binary_df = topic_df.copy()
    topic_binary_df['source_group_binary'] = np.where(topic_binary_df['source_type'] == 'human', 'Human', 'All AI')
    plot_stacked_counts(topic_binary_df, 'source_group_binary', 'lda_topic', title=f'{condition}: fixed LDA topic composition, Human vs All AI (raw proportions; Human n={len(human_idx)}, All AI n={len(ai_idx)})', out_path=pooled_figures_dir / 'lda_topic_composition_human_vs_allai.png', n_map=pooled_group_n)
    lda_binary_tab = pd.crosstab(topic_binary_df['source_group_binary'], topic_binary_df['lda_topic'], normalize='index')
    plot_heatmap(relabel_index_with_n(lda_binary_tab, pooled_group_n), title=f'{condition}: fixed LDA topic proportions, Human vs All AI\nraw proportions; Human n={len(human_idx)}, All AI n={len(ai_idx)}', out_path=pooled_figures_dir / 'lda_topic_heatmap_human_vs_allai.png', fmt='.2f', cmap='rocket_r')

    cluster_binary_df = cluster_df.copy()
    cluster_binary_df['source_group_binary'] = np.where(cluster_binary_df['source_type'] == 'human', 'Human', 'All AI')
    plot_stacked_counts(cluster_binary_df, 'source_group_binary', 'cluster_label', title=f'{condition}: embedding cluster composition, Human vs All AI (raw proportions; Human n={len(human_idx)}, All AI n={len(ai_idx)})', out_path=pooled_figures_dir / 'cluster_composition_human_vs_allai.png', n_map=pooled_group_n)

    nn_binary_df = nn_df.copy()
    nn_binary_df['source_group_binary'] = np.where(nn_binary_df['source_type'] == 'human', 'Human', 'All AI')
    nn_binary_df['global_nn_source_group_binary'] = np.where(nn_binary_df['global_nn_source_type'] == 'human', 'Human', 'All AI')
    plot_stacked_counts(nn_binary_df, 'source_group_binary', 'global_nn_source_group_binary', title=f'{condition}: global 1-NN source composition, Human vs All AI (raw proportions; Human n={len(human_idx)}, All AI n={len(ai_idx)})', out_path=pooled_figures_dir / 'global_nn_source_composition_human_vs_allai.png', n_map=pooled_group_n)
    plot_outlier_prevalence(proposal_metrics_master.assign(source_group=proposal_metrics_master['source_group_binary']), ['is_global_nn_outlier_top10', 'is_global_mean5nn_outlier_top10'], pooled_figures_dir / 'global_nn_outlier_prevalence_human_vs_allai.png', f'{condition}: global NN outlier prevalence, Human vs All AI (raw prevalence; Human n={len(human_idx)}, All AI n={len(ai_idx)})', n_map=pooled_group_n)

    # Save the expanded master after all new columns are available.
    proposal_metrics_master.to_csv(pooled_tables_dir / 'proposal_metrics_master.csv', index=False)
    proposal_metrics_master.to_csv(per_model_tables_dir / 'proposal_metrics_master_per_model.csv', index=False)

    all_metric_tests = []
    for metric in [m for m in MATRIX_DERIVED_METRICS if m in proposal_metrics_master.columns]:
        all_metric_tests.append(matrix_omnibus_group_test(analysis.pairwise_full, group_idx, metric_name=metric, groups=GROUP_ORDER, coords=analysis.umap2d, coord_range=condition_umap_range, condition=condition, family='proposal_level_diversity_and_nn', n_perm=N_PERM))
    for family_name, metrics in [
        ('literature_novelty', NOVELTY_METRICS),
        ('bertopic_region_coverage', [m for m in BERTOPIC_METRICS if m in proposal_metrics_master.columns]),
        ('mesh_term_coverage', [m for m in MESH_METRICS if m in proposal_metrics_master.columns]),
    ]:
        for metric in metrics:
            all_metric_tests.append(kruskal_group_test(proposal_metrics_master, metric, groups=GROUP_ORDER, condition=condition, family=family_name))
    omnibus_df = pd.DataFrame(all_metric_tests)
    omnibus_df['q_value_holm'] = holm_q(omnibus_df['p_value']).to_numpy() if len(omnibus_df) else []
    omnibus_df['significance'] = omnibus_df['q_value_holm'].map(p_stars) if len(omnibus_df) else []
    omnibus_df.to_csv(per_model_tables_dir / 'omnibus_author_group_tests.csv', index=False)

    metric_corr_cols = [c for c in PROPOSAL_LEVEL_METRICS if c in proposal_metrics_master.columns]
    metric_corr_df = proposal_metrics_master[metric_corr_cols].corr(numeric_only=True)
    metric_corr_df.to_csv(pooled_tables_dir / 'proposal_diversity_metric_correlations.csv')
    plot_heatmap(metric_corr_df, title=f'{condition}: proposal metric correlation map', out_path=per_model_figures_dir / 'proposal_metric_correlation_heatmap.png', fmt='.2f', cmap='vlag')

    # Condition-level UMAP with metadata rings and legends outside the graph.
    plot_df = proposal_metrics_master.copy()
    plot_df['umap_x'] = analysis.umap2d[:, 0]
    plot_df['umap_y'] = analysis.umap2d[:, 1]
    fig, ax = plt.subplots(figsize=(8, 7))
    for g in GROUP_ORDER:
        sub = plot_df[plot_df['source_group'] == g]
        ax.scatter(sub['umap_x'], sub['umap_y'], s=58, color=GROUP_PALETTE.get(g, '#808080'), label=label_with_n(g, author_group_n), edgecolor='white', linewidth=0.6, alpha=0.9)
    for flag, color, label, size in [('is_funded_human', '#FF00FF', 'Accepted/Funded Human ring', 105), ('is_top5_AI-ranked', '#FF8C00', 'Top AI-ranked Human ring', 132), ('is_global_mean5nn_outlier_top10', '#00A6D6', 'Mean 5-NN outlier ring', 155)]:
        if flag in plot_df.columns:
            sub = plot_df[plot_df[flag].map(truthy_flag)]
            if len(sub):
                ax.scatter(sub['umap_x'], sub['umap_y'], s=size, facecolors='none', edgecolors=color, linewidths=1.8, label=label)
    ax.set_title(f'{condition}: rephrased proposal-space UMAP with metadata overlays\nraw proposal points; Human n={len(human_idx)}, All AI n={len(ai_idx)}')
    ax.set_xlabel('UMAP-1')
    ax.set_ylabel('UMAP-2')
    ax.legend(loc='upper left', bbox_to_anchor=(1.02, 1.0), frameon=False, fontsize=9)
    save_show(fig, pooled_figures_dir / 'proposal_space_umap_rephrased.png')

    # Existing compact visual summaries, now displayed as well as saved.
    fig, ax = plt.subplots(figsize=(9, 5))
    sns.barplot(data=per_model_summary_df[per_model_summary_df['metric'].isin(PRIMARY_DIVERSITY_METRICS)], x='comparison_group', y='effect_human_minus_group', hue='metric', order=MODEL_ORDER, ax=ax)
    ax.axhline(0, color='black', linewidth=1)
    ax.set_title(f'{condition}: per-model diversity effects (rephrased proposals)\nHuman reference (N={len(human_idx)}); each model x label shows model N')
    ax.set_ylabel('Effect (Human minus model)')
    ax.set_xlabel('Model')
    apply_group_n_xticklabels(ax, MODEL_ORDER, {g: author_group_n[g] for g in MODEL_ORDER}, rotation=20)
    ax.legend(loc='upper left', bbox_to_anchor=(1.02, 1.0), frameon=False)
    save_show(fig, per_model_figures_dir / 'diversity_effects_human_vs_per_model.png')

    fig, ax = plt.subplots(figsize=(8.5, 5))
    pooled_diversity_plot_df = pooled_bootstrap_summary_df[pooled_bootstrap_summary_df['metric'].isin(PRIMARY_DIVERSITY_METRICS)].copy() if len(pooled_bootstrap_summary_df) else pd.DataFrame()
    if len(pooled_diversity_plot_df):
        yerr = np.vstack([
            pooled_diversity_plot_df['effect_human_minus_ai_boot_mean'] - pooled_diversity_plot_df['effect_ci_low'],
            pooled_diversity_plot_df['effect_ci_high'] - pooled_diversity_plot_df['effect_human_minus_ai_boot_mean'],
        ])
        ax.bar(pooled_diversity_plot_df['metric'], pooled_diversity_plot_df['effect_human_minus_ai_boot_mean'], color='#4A90E2', alpha=0.75)
        ax.errorbar(np.arange(len(pooled_diversity_plot_df)), pooled_diversity_plot_df['effect_human_minus_ai_boot_mean'], yerr=yerr, fmt='none', ecolor='black', capsize=4, linewidth=1.2)
    ax.axhline(0, color='black', linewidth=1)
    ax.set_title(f'{condition}: pooled diversity effects, same-size bootstrap\nHuman n={len(human_idx)} vs All AI bootstrap n={len(human_idx)} from {len(ai_idx)}; B={len(bootstrap_ai_idx_samples)}')
    ax.set_ylabel('Effect (Human minus bootstrapped All AI)')
    ax.set_xlabel('Metric')
    ax.tick_params(axis='x', rotation=20)
    save_show(fig, pooled_figures_dir / 'diversity_effects_human_vs_allai.png')

    cross_frame = proposal_metrics_master.copy()
    cross_frame['condition'] = condition
    cross_frame['cross_source_group'] = np.where(cross_frame['source_type'] == 'human', 'Human', f'All-AI-{condition}')
    cross_condition_metric_frames.append(cross_frame)

    condition_results[condition] = {
        'analysis': analysis,
        'proposal_metrics_master': proposal_metrics_master,
        'per_model_summary_df': per_model_summary_df,
        'pooled_summary_df': pooled_summary_df,
        'topic_tests_df': topic_tests_df,
        'cluster_summary': cluster_summary,
        'style_result': style_result,
        'matrix_pairwise_tests': matrix_pairwise_tests,
        'facet_tests_df': facet_tests_df,
        'facet_gradient_df': facet_gradient_df,
        'per_model_tables_dir': per_model_tables_dir,
        'pooled_tables_dir': pooled_tables_dir,
        'per_model_tables_dir': per_model_tables_dir,
        'per_model_figures_dir': per_model_figures_dir,
    }
    print(f'Finished rephrased proposal analysis for {condition}')
```
